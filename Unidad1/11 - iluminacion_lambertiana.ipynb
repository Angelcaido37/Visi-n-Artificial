# Notebook 4 — Iluminación y modelo Lambertiano: por qué una pelota se sombrea

Modelo Lambertiano (reflexión difusa) y el ejemplo de la pelota con linterna.
Regla: **mientras más de frente llega la luz, más brillo** (proporcional a $\cos(\theta)$).

Ejemplo cotidiano: fotos con sombras en cara/objetos.
Ejemplo industrial: inspección donde sombras pueden confundir un algoritmo.


```python
# (1) Importamos librerías/módulos que vamos a usar.
import numpy as np
# (2) Importamos librerías/módulos que vamos a usar.
import matplotlib.pyplot as plt
```

## A) Simular una esfera mate con una luz direccional
Generaremos una esfera 2D (disco) y calcularemos su sombreado Lambertiano.


```python
# 1) Tamaño de la imagen
# (1) Ejecutamos esta instrucción como parte del procedimiento.
H, W = 300, 300

# 2) Coordenadas normalizadas en [-1, 1]
# (2) Definimos/asignamos la variable `y`.
y = np.linspace(-1, 1, H)[:, None]
# (3) Definimos/asignamos la variable `x`.
x = np.linspace(-1, 1, W)[None, :]

# 3) Máscara del disco (x^2 + y^2 <= 1)
# (4) Definimos/asignamos la variable `mask`.
mask = (x**2 + y**2) <= 1

# 4) Para la esfera, la coordenada z se obtiene por z = sqrt(1 - x^2 - y^2)
# (5) Definimos/asignamos la variable `z`.
z = np.zeros_like(x)
# (6) Ejecutamos esta instrucción como parte del procedimiento.
z[mask] = np.sqrt(1 - (x[mask]**2 + y[mask]**2))

# 5) Normal de la superficie en cada punto (x, y, z), normalizada
# (7) Definimos/asignamos la variable `N`.
N = np.dstack([x, y, z])
# (8) Definimos/asignamos la variable `N_norm`.
N_norm = np.linalg.norm(N, axis=2, keepdims=True) + 1e-9
# (9) Definimos/asignamos la variable `N`.
N = N / N_norm

# 6) Definimos la dirección de la luz (vector unitario)
# (10) Definimos/asignamos la variable `L`.
L = np.array([0.6, -0.2, 1.0], dtype=float)
# (11) Definimos/asignamos la variable `L`.
L = L / (np.linalg.norm(L) + 1e-9)

# 7) Lambert: intensidad = max(0, N·L)
# (12) Definimos/asignamos la variable `I`.
I = np.maximum(0, (N @ L))

# 8) Convertimos a 0..255
# (13) Definimos/asignamos la variable `img`.
img = np.zeros((H, W), dtype=np.uint8)
# (14) Convertimos el tipo de dato (por ejemplo a float o uint8).
img[mask] = (255 * I[mask]).astype(np.uint8)

# (15) Creamos una nueva figura para graficar.
plt.figure(figsize=(6,6))
# (16) Mostramos una matriz como imagen.
plt.imshow(img, cmap='gray', vmin=0, vmax=255)
# (17) Agregamos un título a la gráfica.
plt.title('Sombreado Lambertiano (esfera mate)')
# (18) Configuramos/ocultamos ejes para visualizar mejor.
plt.axis('off')
# (19) Renderizamos las gráficas en pantalla.
plt.show()
```

## B) Cambiar el ángulo de luz: “linterna de frente” vs “linterna de lado”
Repetimos con dos direcciones de luz.


```python
# 9) Dos luces: frontal y lateral
# (1) Definimos/asignamos la variable `luces`.
luces = {
    # (2) Ejecutamos esta instrucción como parte del procedimiento.
    'Frontal': np.array([0.0, 0.0, 1.0]),
    # (3) Ejecutamos esta instrucción como parte del procedimiento.
    'Lateral': np.array([1.0, 0.0, 0.4])
# (4) Ejecutamos esta instrucción como parte del procedimiento.
}

# (5) Creamos una nueva figura para graficar.
plt.figure(figsize=(10,5))
# (6) Iniciamos un ciclo para repetir acciones.
for i, (name, Lraw) in enumerate(luces.items(), start=1):
    # (7) Definimos/asignamos la variable `L`.
    L = Lraw / (np.linalg.norm(Lraw) + 1e-9)
    # (8) Definimos/asignamos la variable `I`.
    I = np.maximum(0, (N @ L))
    # (9) Definimos/asignamos la variable `img`.
    img = np.zeros((H, W), dtype=np.uint8)
    # (10) Convertimos el tipo de dato (por ejemplo a float o uint8).
    img[mask] = (255 * I[mask]).astype(np.uint8)
    # (11) Seleccionamos una zona (subplot) dentro de la figura.
    plt.subplot(1,2,i)
    # (12) Mostramos una matriz como imagen.
    plt.imshow(img, cmap='gray', vmin=0, vmax=255)
    # (13) Agregamos un título a la gráfica.
    plt.title(f'Luz {name}')
    # (14) Configuramos/ocultamos ejes para visualizar mejor.
    plt.axis('off')
# (15) Renderizamos las gráficas en pantalla.
plt.show()
```
