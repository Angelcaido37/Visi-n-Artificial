### Objetivo de aprendizaje
Entender que una imagen digital es una **Matriz** (una rejilla) de valores numéricos.

```python
# Primero, importamos nuestras herramientas de trabajo.
# Numpy es como nuestra calculadora científica.
# Matplotlib es nuestro artista para dibujar las gráficas.
import numpy as np
import matplotlib.pyplot as plt

# Esto es para que las imágenes se vean bonitas en el cuaderno
%matplotlib inline
```

### 1. Creando una imagen desde cero (La Matriz)

Vamos a crear una imagen pequeña, de solo 5x5 píxeles. 
Imagina que es una hoja de papel cuadriculado de 5 cuadritos de ancho por 5 de alto.

* Si ponemos un **0**, el cuadrito será **NEGRO** (ausencia de luz).
* Si ponemos un **255**, el cuadrito será **BLANCO** (luz total).
* Los números en medio son diferentes tonos de gris.

```python
# Creamos una matriz (rejilla) de 5x5 con números.
# Fíjate en el patrón: 0 es oscuro, 255 es brillante.
mi_imagen_pequena = np.array([
    [0,   50,  100, 150, 200],
    [50,  100, 150, 200, 250],
    [100, 150, 200, 250, 255],
    [150, 200, 250, 255, 100],
    [200, 250, 255, 100, 50]
])

print("Así se ve la imagen para la computadora (Números):")
print(mi_imagen_pequena)
```

### 2. Visualizando la Matriz

Ahora, vamos a pedirle a la computadora que pinte esos números como si fueran intensidades de luz. 
Recuerda: **Píxel** significa "Picture Element" (Elemento de imagen). Es cada celda de esta matriz.

```python
# Vamos a mostrar la matriz como una imagen en escala de grises
plt.figure(figsize=(5, 5))

# 'imshow' significa 'image show' (mostrar imagen)
# 'cmap="gray"' le dice que use una paleta de grises (blanco y negro)
# 'vmin=0, vmax=255' asegura que entienda que 0 es negro y 255 blanco
plt.imshow(mi_imagen_pequena, cmap='gray', vmin=0, vmax=255)

# Agregamos los valores numéricos encima para ver la correspondencia
for i in range(5):
    for j in range(5):
        # Escribimos el número en el centro de cada pixel
        plt.text(j, i, mi_imagen_pequena[i, j], 
                 ha="center", va="center", color="red")

plt.title("Tu primera imagen digital: Una Matriz")
plt.axis('off') # Quitamos los ejes para que se vea más limpio
plt.show()
```

### Conclusión del ejercicio

¡Felicidades! Acabas de ver el concepto más importante de la Visión Artificial:

> **Una imagen $f(x, y)$ es una función matemática discretizada.**

Cuando la cámara toma una foto, lo que hace es medir cuánta luz llega a cada puntito del sensor y guardar ese valor en una celda de esta matriz. ¡No hay magia, son matemáticas!
