# 💡 Unidad 1: Formación Radiométrica
## Tema: El Modelo de Iluminación ($f = i \cdot r$)

¡Hola! 👋 Este es el último concepto clave para dominar la Unidad 1.

Hasta ahora hemos visto píxeles y geometría, pero... **¿de dónde sale el valor de gris de un píxel?**

La teoría nos dice que una imagen $f(x,y)$ es el resultado de MULTIPLICAR dos cosas:
1.  **Reflectancia $r(x,y)$:** Es la "pintura" del objeto. (Ej: Un papel es blanco, el carbón es negro). Esto va de 0 a 1.
2.  **Iluminación $i(x,y)$:** Es la luz que cae sobre el objeto. (Ej: Una lámpara potente, una sombra).

$$ f(x,y) = i(x,y) \cdot r(x,y) $$

Vamos a simular esto para entender por qué es tan difícil para una computadora ver en la oscuridad.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

### 1. Creando la Realidad de los Objetos (Reflectancia)

Imagina que tenemos una pared con texturas: ladrillos claros y oscuros. Esto es la **Reflectancia ($r$)**. Es una propiedad física del material y NO cambia aunque apagues la luz.

```python
# Creamos una textura de 'tablero de ajedrez' para simular materiales distintos
reflectancia = np.zeros((200, 200))

# Cuadros blancos (Reflectancia alta = 1.0) y negros (Reflectancia baja = 0.2)
# Usamos 0.2 en lugar de 0 porque en la vida real nada es 'perfectamente' negro.
tamano_cuadro = 40
for y in range(0, 200, tamano_cuadro):
    for x in range(0, 200, tamano_cuadro):
        if (x // tamano_cuadro + y // tamano_cuadro) % 2 == 0:
            reflectancia[y:y+tamano_cuadro, x:x+tamano_cuadro] = 1.0 # Blanco perfecto
        else:
            reflectancia[y:y+tamano_cuadro, x:x+tamano_cuadro] = 0.2 # Gris oscuro

plt.figure(figsize=(4, 4))
plt.title("Reflectancia $r(x,y)$ \n(La verdad del objeto)")
plt.imshow(reflectancia, cmap='gray', vmin=0, vmax=1)
plt.colorbar()
plt.axis('off')
plt.show()
```

### 2. Creando la Luz (Iluminación)

Ahora imagina que hay una lámpara a la izquierda de la pared. La luz no llega igual a todos lados; se va perdiendo fuerza mientras se aleja. Esto es la **Iluminación ($i$)**.

```python
# Creamos un degradado de luz (fuerte a la izquierda, oscuro a la derecha)
x = np.linspace(0, 1, 200)
y = np.linspace(0, 1, 200)
X, Y = np.meshgrid(x, y)

# Simulación de luz: Una rampa que va de intensidad 250 a intensidad 50
iluminacion = 250 * (1 - X) + 50

plt.figure(figsize=(4, 4))
plt.title("Iluminación $i(x,y)$ \n(La Luz de la lámpara)")
plt.imshow(iluminacion, cmap='gray', vmin=0, vmax=255)
plt.colorbar()
plt.axis('off')
plt.show()
```

### 3. La Imagen Resultante (Lo que ve la cámara)

Aquí ocurre la magia de la física. La cámara captura el producto de ambas.
$$ f = i \times r $$

```python
# Multiplicación elemento a elemento
imagen_capturada = iluminacion * reflectancia

plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.title("Reflectancia ($r$)")
plt.imshow(reflectancia, cmap='gray', vmin=0, vmax=1)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Iluminación ($i$)")
plt.imshow(iluminacion, cmap='gray', vmin=0, vmax=255)
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Imagen Final ($f = i \cdot r$)")
plt.imshow(imagen_capturada, cmap='gray', vmin=0, vmax=255)
plt.axis('off')

plt.show()
```

### Reflexión Importante (El Problema de la Visión)

Mira la "Imagen Final" a la derecha.

Fíjate en los **cuadros blancos** del lado DERECHO (donde está oscuro). 
Sus valores numéricos son **más bajos** (más oscuros) que los **cuadros negros** del lado IZQUIERDO (donde hay mucha luz).

Esto es un problema grave para las computadoras: **¿Cómo sabe el robot que el cuadro de la derecha es blanco si su número dice que es oscuro?**

Esto nos lleva al concepto de **"Problema Mal Planteado" (Ill-posed problem)** mencionado en tu Unidad 1: Es fácil calcular la imagen si tienes la luz y el objeto, pero es muy difícil recuperar el objeto si solo tienes la imagen.

```python
# Demostración numérica del problema
valor_negro_con_luz = imagen_capturada[10, 10]   # Esquina superior izquierda (Material oscuro, Luz fuerte)
valor_blanco_sin_luz = imagen_capturada[10, 190] # Esquina superior derecha (Material claro, Luz pobre)

print(f"Valor del píxel 'NEGRO' con mucha luz: {valor_negro_con_luz:.2f}")
print(f"Valor del píxel 'BLANCO' con poca luz: {valor_blanco_sin_luz:.2f}")

if valor_negro_con_luz > valor_blanco_sin_luz:
    print("\nCONCLUSIÓN: ¡La computadora está confundida! Cree que el negro es más brillante que el blanco.")
```
