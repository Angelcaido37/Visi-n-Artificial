## 1. Resta de Fondo (MOG2)

Este algoritmo aprende qué partes de la escena son estáticas (el fondo) y resalta solo lo que cambia (el frente).

```python
import cv2

cap = cv2.VideoCapture('trafico_video.mp4')

# Crear el objeto de sustracción de fondo
fgbg = cv2.createBackgroundSubtractorMOG2(history=500, varThreshold=50, detectShadows=True)

while True:
    ret, frame = cap.read()
    if not ret: break
    
    # Aplicar la máscara de movimiento
    fgmask = fgbg.apply(frame)
    
    # Limpiar ruido (Erosión/Dilatación que aprendiste en Semana 5)
    fgmask = cv2.medianBlur(fgmask, 5)
    
    cv2.imshow('Movimiento Detectado', fgmask)
    if cv2.waitKey(30) & 0xFF == 27: break

cap.release()
cv2.destroyAllWindows()
```
