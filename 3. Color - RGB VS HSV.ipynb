## Tema: Espacios de Color (RGB vs HSV)

¡Bienvenido a la clase de arte digital! 🎨

Hasta ahora hemos visto imágenes en gris, pero el mundo es a color. 
En la presentación aprendimos que hay dos formas principales de describir el color:

1.  **RGB (Red, Green, Blue):** Así piensa el hardware (tu monitor, tu cámara). Mezcla luces.
2.  **HSV (Hue, Saturation, Value):** Así pensamos los humanos. "¿Qué color es? (Hue), ¿Qué tan pálido es? (Saturation), ¿Qué tan brillante es? (Value)".

Vamos a ver por qué el HSV es el favorito de los ingenieros de Visión Artificial.

```python
import numpy as np
import matplotlib.pyplot as plt
# Usaremos OpenCV, la librería más famosa de Visión Artificial
# Si estuvieras en tu compu, se instala con: pip install opencv-python
import cv2
```

### 1. Creando una imagen a color

Vamos a crear una imagen sintética con tres cuadros: 
* Uno **ROJO** puro.
* Uno **VERDE** puro.
* Uno **AZUL** puro.

Recuerda: Una imagen a color son TRES matrices apiladas (como un sándwich de 3 capas).

```python
# Creamos una imagen negra de 100x300 pixeles con 3 canales (RGB)
imagen_color = np.zeros((100, 300, 3), dtype=np.uint8)

# Pintamos el primer tercio de ROJO (Canal R=255, G=0, B=0)
imagen_color[:, 0:100] = [255, 0, 0]

# Pintamos el segundo tercio de VERDE (Canal R=0, G=255, B=0)
imagen_color[:, 100:200] = [0, 255, 0]

# Pintamos el último tercio de AZUL (Canal R=0, G=0, B=255)
imagen_color[:, 200:300] = [0, 0, 255]

plt.figure(figsize=(6, 2))
plt.title("Nuestra Imagen RGB Sintética")
plt.imshow(imagen_color)
plt.axis('off')
plt.show()
```

### 2. Separando canales (La visión de la máquina)

Si separamos las capas, veremos que el "Rojo" es brillante en la primera capa y negro en las otras. Es difícil separar colores si hay variaciones de luz.

```python
# Separamos los canales
R = imagen_color[:, :, 0]
G = imagen_color[:, :, 1]
B = imagen_color[:, :, 2]

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(12, 3))

ax1.set_title("Canal Rojo (R)")
ax1.imshow(R, cmap='gray') # Se ve blanco donde hay rojo

ax2.set_title("Canal Verde (G)")
ax2.imshow(G, cmap='gray') # Se ve blanco donde hay verde

ax3.set_title("Canal Azul (B)")
ax3.imshow(B, cmap='gray') # Se ve blanco donde hay azul

plt.show()
```

### 3. La magia del HSV (Hue, Saturation, Value)

Ahora vamos a convertir la imagen a HSV. 

* **H (Hue - Matiz):** Nos dice QUÉ color es (un ángulo en el círculo cromático).
* **S (Saturation):** Nos dice qué tan puro es.
* **V (Value):** Nos dice qué tanta luz tiene.

**¿Por qué es útil?** Si quisieras detectar una pelota roja que está mitad en el sol y mitad en la sombra:
* En RGB fallarías, porque la parte en sombra tiene valores RGB muy distintos a la parte al sol.
* En HSV, ¡ambas partes tienen el mismo **Hue** (Matiz)! Solo cambia el brillo (V).

```python
# Convertimos de RGB a HSV usando OpenCV
imagen_hsv = cv2.cvtColor(imagen_color, cv2.COLOR_RGB2HSV)

# Extraemos el canal Hue (Matiz)
H = imagen_hsv[:, :, 0]

plt.figure(figsize=(6, 3))
plt.title("Canal Hue (El 'ADN' del color)")
# Usamos un mapa de colores 'hsv' para que tenga sentido visual
plt.imshow(H, cmap='hsv') 
plt.colorbar(label='Valor del Matiz')
plt.axis('off')
plt.show()
```

### Conclusión

Fíjate en la barra de colores de la derecha. Cada color tiene un valor numérico único sin importar su brillo. 

* El Rojo está cerca del valor 0 (o 180 en OpenCV).
* El Verde está por el 60.
* El Azul está por el 120.

¡Así es como los robots pueden seguir una pelota roja o detectar un limón maduro (amarillo) de uno verde, tal como vimos en las diapositivas!
